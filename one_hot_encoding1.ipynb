```python
import pandas as pd
data = pd.read_csv("carprices.csv")
data
```

```python
dummies = pd.get_dummies(data['Car Model'])
dummies
```

```python
merged_data = pd.concat([data, dummies], axis = 'columns')
merged_data
```

```python
final_data = merged_data.drop(['Car Model','Audi A5'],axis = 'columns')
final_data
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
X = final_data.drop(['Sell Price($)'],axis = 'columns')
X
```

```python
Y = final_data['Sell Price($)']
Y
```

```python
model.fit(X, Y)
```

```python
model.predict([[45000,4,0,1]])
```
