```python
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
data = pd.read_csv('insurance_data.csv')
data.head(6)
```

```python
plt.scatter(data.age, data.bought_insurance, marker='+', color='red');
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data[['age']], data.bought_insurance, train_size=0.8)
X_test
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
X_test
```

```python
model.predict(X_test)
```

```python
model.score(X_test, y_test)
# model.score will give the accuracy of the model, if model is perfect it will give 1.0,
# if they are any issues it will give value like 0. something
```

```python
model.predict_proba(X_test)
# this will provide the probability of weather they will take insurance or not, based on their age.
```

```python
# DEFINING SIGMOID FUNCTION
import math
def sigmoid(x):
  return 1 / (1 + math.exp(-x))

def prediction_function(age):
  z = 0.156 * age - 6.52
  y = sigmoid(z)
  return y
age = 30
prediction_function(age)
# 0.13705129257545967 which is less that 0.5 hence the person with age 30 will
# not take insurance.
age = 50
prediction_function(age)
# 0.7824497764231124 which is grater than 0.5 hence the perseon with age 50 will
# take insurance.
```
